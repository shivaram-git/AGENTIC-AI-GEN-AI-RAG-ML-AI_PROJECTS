## 4 Logical Operator

Logical operators are used to **combine multiple conditions**.

Python provides three logical operators:

| Operator | Meaning |
|:---:|---|
| `and` | Returns `True` when **both conditions are True** |
| `or` | Returns `True` when **at least one condition is True** |
| `not` | **Reverses** the result |

```python
a = 5
b = 4
```

```python
a < 8 and b < 5
```

```python
a > 8 and b < 5
```

```python
a > 8 or b < 5
```

```python
a < 8 and b < 5
```

```python
a > 8 and b > 5
```

```python
x = False
x
```

```python
not x
```

```python
y = True
y
```

```python
not y
```

```python
age = 25

# AND
print(age > 18 and age < 30)

# OR
print(age < 18 or age > 20)

# NOT
print(not(age > 18))
```



## 5 Unary Operators

A unary operator works with **only one value or variable**.

The common unary operators are:

| Operator | Meaning | Example | Result |
|:---:|---|---:|---:|
| `+` | Positive | `+10` | `10` |
| `-` | Negative | `-10` | `-10` |

###  Simple Meaning

> **Unary = One operand**

Unlike arithmetic operations such as `10 + 5`, which use two values, a unary operator works with only one value.

```python
x = 10

print("Positive:", +x)
print("Negative:", -x)
```
