### 1. if

if executes a block of code when the condition is True

```python
age = 20

if age >= 18:
    print("Eligible")
```

### 2. Multiple if Statements

When we use multiple if statements, each condition is checked independently.

```python
age = 20

if age >= 18:
    print("Adult")

if age >= 21:
    print("Can drink")
```

### 3. if-else

else is used when there are two possible outcomes.

```python
age = 20

if age >= 18:
    print("Eligible")
else:
    print("Not eligible")
```

### 4. if-elif-else

elif means "else if".

It is used when we have more than two possible conditions.

Use if-elif-else when you have multiple mutually exclusive choices.

Because Python checks the conditions one by one and stops when it finds the first True condition.


```python
marks = 75

if marks >= 90:
    print("A")
elif marks >= 75:
    print("B")
elif marks >= 60:
    print("C")
else:
    print("Fail")
```

### What is Nested if?

A nested if means putting one if statement inside another if statement.

In simple words:

First check one condition. If it is True, then check another condition.

```python
registered = True
has_id = True

if registered:
    print("Registration verified")

    if has_id:
        print("You can enter the exam hall")
```

```python
age = 20
has_id = False

if age >= 18:
    if has_id:
        print("Entry allowed")
    else:
        print("ID required")
else:
    print("You are under 18")
```
