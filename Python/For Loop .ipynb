BREAK STATEMNT - if you apply break statment in a loop then it will end the loop 

Pass = skips block of code( function, class etc)

Continue= skips 1 step/iteration during loop 

Break= jumps out of the function/loop

# **break**

```python
for i in range(1,11):
    if i == 5:
        break
    print(i)
```

# **continue**

```python
for i in range(1,8):
    if i == 2:
        continue
    print(i)
```

```python
for i in range(1,11):
    if i == 6:
        continue
    print('hello',i)
```

# for else
 For|Else in python
 
- In other language for else not supportable but in python it is supportable

eg- lets print the number from 1- 20 & we dont want print number which is divisible by 5


```python
nums = [12,15,18,21,26]

for num in nums:
    if num % 5 == 0:
        print(num)  
```

```python
nums = [12,14,18,21,25,30,35]

for num in nums:
    if num % 5 == 0:
        print(num)  
```

```python
nums = [12,14,18,21,25,20]

for num in nums:
    if num % 5 == 0:
        print(num)  
```

```python
nums = [12,14,18,21,20,25]

for num in nums:
    if num % 5 == 0:
        print(num)  
        break
```

```python
nums = [7,14,18,21,23,27] #there is no number which is divisible by 5 we got output as blank

for num in nums:
    if num % 5 == 0:
        print(num)  
        break
```

```python
nums = [7,14,18,21,23,27,29] #hear there is no number which is divisible by 5 we got output as blank

for num in nums:
    if num % 5 == 0:
        print(num)  
        break
    else:
        print('Number Not Found') #every iteration it cheking condition        
```

```python
nums = [7,14,18,21,23,27] 

for num in nums:
    if num % 5 == 0:
        print(num)  
        break
else:
    print('number not found')
```

# prime number

```python
num = 19

for i in range(2,num):
    if num % i == 0:
        print('Not prime Number')
        break
else:
    print('Prime Number')
```
