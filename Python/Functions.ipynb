```python
def greet():
    return 'Hello'

def add_sub(x,y):
    c  = x + y 
    d = x - y

    return c,d

r = greet()
r1,r2 = add_sub(10,2)
print(r)
print(r1)
print(r2)

print(type(r))
print(type(r1))
print(type(r2))
```

```python

def add_sub_mul(x,y):
    c  = x + y 
    d = x - y
    e = x * y

    return c,d,e


r1,r2,r3 = add_sub_mul(10,2)

print(r1)
print(r2)
print(r3)


print(type(r1))
print(type(r2))
print(type(r3))
```
