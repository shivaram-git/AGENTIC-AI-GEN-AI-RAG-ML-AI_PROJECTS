### iterative :
Same tasks will repeatedly

```python
i = 1                                   # Initializing
while i <= 5:                           # Condition
    print("Data Science",i)
    i += 1                              # Increment
```

```python
i = 1
while i <= 5:
    print("Love you Shivaram ❤️")
    i += 1
```

```python
i = 5
while i > 0:
    print('Data Science',i)
    i = i-1
    
```

```python
i = 1

while i <= 5:
    print("Data Science")

    j = 1
    while j <= 4:
        print("Technology")
        j += 1
    i = i + 1

    print()

```

```python
i = 1

while i <= 5:
    print("Data Science",end=' ')

    j = 1
    while j <= 4:
        print("Technology",end=' ')
        j += 1
    i = i + 1

    print()

```
