## 1. What is Matplotlib?

Matplotlib is a Python library used for data visualization.

It is mainly used to create:

- Line plots
- Bar charts
- Scatter plots
- Histograms
- Pie charts
- Box plots
- Area plots
- 3D plots
- Subplots
- Customized visualizations

Matplotlib is one of the most important visualization libraries used in Python Data Analysis.

### Why do we use Matplotlib?

Matplotlib helps us convert numerical data into visual information.

For example:

- Sales data → Bar chart
- Monthly trend → Line chart
- Distribution → Histogram
- Relationship between variables → Scatter plot
- Outliers → Box plot

## 16. Grid

The `plt.grid()` function adds grid lines to a plot.

Grid lines make it easier to read values from the graph.

````python
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]

plt.plot(x, y)

plt.grid()

plt.show()
````

### Customize Grid

````python
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]

plt.plot(x, y)

plt.grid(
    linestyle="--",
    linewidth=0.5,
    alpha=0.7
)

plt.show()
````

## 18. Legend

A legend explains what each line or data series represents.

We use:

```python
label="..."

````python

months = [1, 2, 3, 4, 5]

sales = [100, 150, 180, 220, 300]

plt.plot(
    months,
    sales,
    label="Sales"
)

plt.legend()

plt.show()
````

## 19. Multiple Lines

We can plot multiple data series on the same graph.

````python
months = [1, 2, 3, 4, 5]

sales_2025 = [100, 150, 180, 220, 300]
sales_2026 = [120, 170, 200, 250, 350]

plt.plot(
    months,
    sales_2025,
    label="2025"
)

plt.plot(
    months,
    sales_2026,
    label="2026"
)

plt.title("Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.legend()
plt.grid()

plt.show()
````



# 26. Scatter Plot

A scatter plot is used to visualize the relationship between two numerical variables.

Function:

```python
plt.scatter(x, y)

````python
height = [150, 155, 160, 165, 170, 175, 180]
weight = [50, 55, 58, 62, 68, 72, 78]

plt.scatter(height, weight)

plt.title("Height vs Weight")
plt.xlabel("Height")
plt.ylabel("Weight")

plt.show()
````

````python
height = [150, 155, 160, 165, 170, 175, 180]
weight = [50, 55, 58, 62, 68, 72, 78]

plt.scatter(
    height,
    weight,
    s=100,
    alpha=0.8,
    marker="o"
)

plt.title("Height vs Weight")
plt.xlabel("Height")
plt.ylabel("Weight")


plt.show()
````

# 29. Histogram

A histogram shows the distribution of numerical data.

It divides data into intervals called **bins**.

Function:

```python
plt.hist()

````python

marks = [
    45, 50, 52, 55, 58,
    60, 62, 65, 67, 70,
    72, 75, 78, 80, 82,
    85, 88, 90, 92, 95
]

plt.hist(marks)

plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
````

## 30. Histogram Bins

The `bins` parameter controls how the data is divided into intervals.

````python
marks = [
    45, 50, 52, 55, 58,
    60, 62, 65, 67, 70,
    72, 75, 78, 80, 82,
    85, 88, 90, 92, 95
]

plt.hist(marks, bins=10)

plt.title("Marks Distribution")

plt.show()
````

# 32. Area Plot

We can use `fill_between()` to create an area plot.

Area plots are useful for showing the magnitude of values over an interval.

````python
months = [1, 2, 3, 4, 5, 6]

sales = [100, 150, 130, 180, 220, 250]

plt.plot(months, sales)

plt.fill_between(months, sales, alpha=0.3)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
````

# 33. Subplots

Subplots allow us to create multiple plots inside one Figure.

Syntax:

```python
fig, axes = plt.subplots(rows, columns)

````python

fig, axes = plt.subplots(2, 2)

axes[0, 0].plot([1, 2, 3], [10, 20, 30])
axes[0, 1].bar(["A", "B", "C"], [10, 20, 30])
axes[1, 0].scatter([1, 2, 3], [30, 20, 40])
axes[1, 1].hist([10, 20, 20, 30, 30, 30])

plt.show()
````

````python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0, 0].plot([1, 2, 3], [10, 20, 30])
axes[0, 0].set_title("Line Plot")

axes[0, 1].bar(["A", "B", "C"], [10, 20, 30])
axes[0, 1].set_title("Bar Chart")

axes[1, 0].scatter([1, 2, 3], [30, 20, 40])
axes[1, 0].set_title("Scatter Plot")

axes[1, 1].hist([10, 20, 20, 30, 30, 30])
axes[1, 1].set_title("Histogram")

plt.tight_layout()

plt.show()
````

## 34. tight_layout()

When multiple plots are present, labels and titles may overlap.

`plt.tight_layout()` automatically adjusts spacing between subplots.

````python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0, 0].plot([1, 2, 3], [10, 20, 30])
axes[0, 1].bar(["A", "B", "C"], [10, 20, 30])
axes[1, 0].scatter([1, 2, 3], [30, 20, 40])
axes[1, 1].hist([10, 20, 20, 30, 30, 30])

plt.tight_layout()

plt.show()
````

## Object Oriented API

````python
fig = plt.figure()

x2 = np.linspace(0, 5, 10)
y2 = x2 ** 2

axes = fig.add_axes([0.1, 0.1, 0.8, 0.8])

axes.plot(x2, y2, 'r')

axes.set_xlabel('x2')
axes.set_ylabel('y2')
axes.set_title('title');
````
