# Univariate Analysis with seaborn

```python
# plot the graph using 1 variable is called UNivarate Analysis

vis1 = sns.distplot(df['InternetUsers'])
```

```python
vis2 = sns.displot(df['InternetUsers'],bins = 10)
```

# Bivariate Analysis
**plot the graph using 2 variable is called Bivariate Analysis**

```python
vis4 = sns.boxplot(data = df,x = 'IncomeGroup',y='BirthRate')
```

```python
# outlier statistics outlier is the datapoint which is very far from the other obervation
# outlier also called as Anomoaly Detetction

# if we found outlier in the dataset we introduce probability function()
```

```python
vis5 = sns.lmplot(data = df , x = 'InternetUsers', y = 'BirthRate')
```

```python
vis5 = sns.lmplot(data = df , x = 'InternetUsers', y = 'BirthRate',fit_reg = False)
```

```python
vis6 = sns.lmplot(data = df , x = 'InternetUsers', y = 'BirthRate',fit_reg = False , hue = 'IncomeGroup')
```

```python
vis7 = sns.lmplot(data = df , x = 'InternetUsers', y = 'BirthRate',fit_reg = True , hue = 'IncomeGroup')
```
